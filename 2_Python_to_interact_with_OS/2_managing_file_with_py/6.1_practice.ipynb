```python
import csv

f = open('csv_file.txt')
# It is parsing through the file
# and reading the contents
# using the csv.reader() method
# The csv.reader() method takes a file object as an argument
csv_f = csv.reader((f))
# Iterating through the csv_f vairable
for row in csv_f:
    # Printing the contents of the file
    # unpacking the values into variables
    name, phone, role = row
    print("Name: {}, Phone: {}, Role: {}".format(name, phone, role))
f.close()
```

```python
# Reading and Writing CSV files with Dictionaries.
# DictReader turns each row in a csv file into a dictionary
# with the keys being the first row of the csv file

# We can user DictWriter to generate a csv file from a list of dictionaries

with open('software.csv') as software:
    reader = csv.DictReader(software)
    # Earlier we were storing rows into one variable to let it print
    for row in reader:
        print(("{} has {} users").format(row["name"], row["users"]))
```

```python
# Now generating a csv from a list of Dictionaries.
users = [
    {"name": "Sol Mansi", "username": "solm", "department": "IT Infrastructure"},
    {"name": "Martha Stewart", "username": "martha", "department": "DevOps"},
    {"name": "Bill Gates", "username": "billg", "department": "Development"}
]

keys = ["name", "username", "department"]

with open('by_department.csv', 'w') as f:
    writer = csv.DictWriter(f, fieldnames=keys)
    # The writeheader() method will write the first row of the csv file
    # with the keys of the dictionary
    writer.writeheader()
    # The writerows() method will write the rows of the csv file
    # with the values of the dictionary
    writer.writerows(users)
# The above code will generate a csv file with the following contents:
```

```python
import csv

def read_employees(csv_file_location):
    with open(csv_file_location) as f:
        reader = csv.DictReader(f)
        employee_list = []
        for row in reader:
            employee_list.append(row)
    return employee_list
list_of_employees = read_employees('/Users/vishv/Coursera_IT_Automation/2_Python_with_OS/2_managing_file_with_py/employees.csv')
print(list_of_employees)


# Processing the List of Dictionaries that we have just prepared.)
def process_data(employee_list):
    department_list = []
    for employee_data in employee_list:
        department = employee_data.get(' Department')
        if department:  # Check if the department is empty or not or None _----> If it is empty it will return True and run the follwing line of code.
            cleaned_department = department.strip() # Remove leading/trailing whitespace
        if cleaned_department not in department_list:
            department_list.append(cleaned_department)
    return department_list
department_list = process_data(list_of_employees)
print(department_list)
```
