```python
# File size retriever using OS module.
import os
size = os.path.getsize('helm.tar.gz')

actual_size = size / 1024 / 1024 
print(f"Size of the file is: {actual_size:.2f} MB")
```

```python
# This will provide a unix timestamp for the file
import os
import time
stamp = os.path.getmtime('helm.tar.gz')
print(f"Last modified time of the file is: {stamp:.2f}")
```

```python
import datetime
import os
# This will convert the timestamp to a human-readable format
timestamp = os.path.getmtime('helm.tar.gz')
dt_object = datetime.datetime.fromtimestamp(timestamp)
date_string = dt_object.strftime('%Y-%m-%d %H:%M:%S')
print(date_string)
# This will print the date in a human-readable format
# year, month, day, hour, minute, microseconds
```

```python
# This will generate an absolute path for the file

os.path.abspath('helm.tar.gz')
```
