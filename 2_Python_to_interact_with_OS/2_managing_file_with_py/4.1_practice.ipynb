```python
# checking if the path exists or not
import os
path = 'spider.txt'
path2 = 'example.txt'
if os.path.exists(path):
    print(f"The path '{path}' exists.")
else:
    print(f"The path '{path}' does not exist.")

if os.path.exists(path2):
    print(f"The path '{path2}' exists")
else:
    print(f"The path '{path2} does not exists'")

# os.path.join --> Joins the path components using the correct directory separator and append '<filename>'
file_path = os.path.join(os.getcwd(), 'spider.txt')
print(file_path)
```

```python
# checking if a Path is a file or directory
import os
path = 'area.py'
if os.path.isfile(path):
    print(f"'{path}' is a file.")
elif os.path.isdir(path):
    print(f"'{path}' is a directory.")
else:
    print(f"'{path}' does not exists.")
```

```python
new_guests = ["Sammy", "Yogesh", "Bla-Bla"]
with open ("guests.txt", 'a') as guests:
    for i in new_guests:
        guests.write(i+"\n") # It is there to insert new line character
```

```python
# Let's remove the guests name from the list
checked_out = ["Dave", "Bob", "Alice"]
temp_list =[]

with open("guests.txt", 'r') as guests:
    for g in guests:
        temp_list.append(g.strip())
    
with open("guests.txt", 'r+') as guests:
    for name in temp_list:
        if name not in checked_out:
            guests.write(name + "\n")
```
