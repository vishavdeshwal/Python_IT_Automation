```python
# To check whether there is punctuation in the string or not

import re
def check_punctuation (text):
    # It will see if the raw string has any of the characters mentioned in the closed brackets.
  result = re.search(r"[.?!]", text)
  return result != None

print(check_punctuation("This is a sentence that ends with a period.")) # True
print(check_punctuation("This is a sentence fragment without a period")) # False
print(check_punctuation("Aren't regular expressions awesome?")) # True
print(check_punctuation("Wow! We're really picking up some steam now!")) # True
print(check_punctuation("End of the line")) # False


# This ^ (carrot) symbol inside the brackets means "not" in the regex ----> any character in the bracket should not match.
print(re.search(r"[^a-zA-Z ]", "This is a sentence with spaces."))

# This pipe symbol | means "or" in the regex ----> either of the two should match.
# If we provide both the options, then it will match the first match it find, it can be either of the two.
print(re.search(r"cat|dog", "I like dogs and cats"))

# But if we want to match both the options, then we can use findall() method.
print(re.findall(r"cat|dog", "I like dogs and catds"))
```

```python
# Escaping Charaters
import re

print(re.search(r".com", "welcome"))
print(re.search(r"\.com", "welcome"))
print(re.search(r"\.com", "mydomain.com"))

# \w means matching any alphanumeric character including underscore.
# * after \w means that the character can be repeated zero or more times up until the end of the string or the first space.
print(re.search(r"\w*", "This is an example"))
print(re.search(r"\w*", "And_this_is_another_example"))
```

```python
# ^ and $ if used like this--------> It means that we specifically wants our string to start and ends with the denoted character.
print(re.search(r"^A.*a$", "Azerbejaan"))
print(re.search(r"^A.*a", "Azerbejaan"))


# $ will ensure that till the end of the string, the pattern should match, if anything in the pattern is not there, it will say None
# however, if * is there, it will go to the last character of the string that is matching it.
pattern = r"^[a-zA-Z_][a-zA-Z0-9_ ]*$"

print(re.search(pattern, "This isn't a valid string"))
```

```python
import re
def check_web_address(text):
    # * means that the character can be repeated zero or more times
    # . means that there can be any character in between
  pattern = f"^[a-zA-Z0-9_.-]*.[a-zA-Z]$"
  result = re.search(pattern, text)
  return result != None

print(check_web_address("gmail.com")) # True
print(check_web_address("www@google")) # False
print(check_web_address("www.Coursera.org")) # True
print(check_web_address("web-address.com/homepage")) # False
print(check_web_address("My_Favorite-Blog.US")) # True
```

```python
# Time calculation and matching pattern

import re
def check_time(text):
    #\s? means that there can be zero or one space or special character (space, tab, new line) and ? means it can be there or it cannot be.
  pattern = r"^1[0-2]|[0-9]:[0-5][0-9]\s?[aApP][mM]$"
  result = re.search(pattern, text)
  return result != None

print(check_time("12:45pm")) # True
print(check_time("9:59 AM")) # True
print(check_time("6:60am")) # False
print(check_time("five o'clock")) # False
print(check_time("6:02 am")) # True
print(check_time("6:02km")) # False
```

```python
import re
def contains_acronym(text):
  pattern = r"[(][a-zA-Z0-9].*[)]"
  result = re.search(pattern, text)
  return result != None

print(contains_acronym("Instant messaging (IM) is a set of communication technologies used for text-based communication")) # True
print(contains_acronym("American Standard Code for Information Interchange (ASCII) is a character encoding standard for electronic communication")) # True
print(contains_acronym("Please do NOT enter without permission!")) # False
print(contains_acronym("PostScript is a fourth-generation programming language (4GL)")) # True
print(contains_acronym("Have fun using a self-contained underwater breathing apparatus (Scuba)!")) # True
```

```python
# zip code

import re

def correct_function(text):
  result = re.search(r". [0-9 ]{5}", text)  # Corrected regex pattern with space
  return result is not None

def check_zip_code(text):
  return correct_function(text)  # Call the correct_function

# Call the check_zip_code function with test cases
print(check_zip_code("The zip codes for New York are 10001 thru 11104."))  # True
print(check_zip_code("90210 is a TV show"))  # False (no space before 90210)
print(check_zip_code("Their address is: 123 Main Street, Anytown, AZ 85258-0001."))  # True
print(check_zip_code("The Parliament of Canada is at 111 Wellington St, Ottawa, ON K1A0A9."))  # False
```
