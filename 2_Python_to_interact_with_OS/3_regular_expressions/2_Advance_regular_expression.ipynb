```python
# Capturing Groups
import re
result = re.search(r"^(\w*), (\w*)$", "Lovelace, Ada")
# \w = It will match any alphanumeric character including underscore.
# * after \w means that the character can be repeated zero or more times up until the end of the string or the first space.

print(result)
# because it has two groups, we will have the tupples of two groups
print(result.groups())
# This will be the whole String
print(result[0])

# This will capture the first capturing group = Lovelace
print(result[1])

# This will capture the second capturing group = Ada
print(result[2])
```

```python
import re
def rearrange_name(name):
  result = re.search(r"^([\w\-\.]+),\s+([\w\s\.-]*)$", name)
  if result == None:
    return name
  return "{} {}".format(result[2], result[1])

name=rearrange_name("Kennedy, John F.")
print(name)
```

```python
# Developing function based on the capturing groups
import re
print(re.search(r"[a-zA-Z]{5}", "a scary ghost appeared")) # This will check first five characters
# If we include space then it will read all five charcters from start
print(re.search(r"[a-zA-Z ]{5}", "a scary ghost appeared"))

# This will find all the five charater words irrespective and provide the list of it.
print(re.findall(r"[a-zA-Z]{5}", "a scary ghost appeared"))

# This will find all the five character words only, it will not clip the five words from a longer word like previous one
#/b denotes that it is a word boundary, it will only look for specific long character which is defined
print(re.findall(r"\b[a-zA-Z]{5}\b", "A scary ghost appeared"))

# It will find all the five to ten character words as mentioned in the length range
print(re.findall(r"\w{5,10}", "I really like strawberries"))

# This will find all the five or more character words as there is not limit set after {5,}
print(re.findall(r"\w{5,}", "I really like strawberries"))


```

```python
import re
def long_words(text):
  pattern = r"[\w]{7,}"
  result = re.findall(pattern, text)
  return result

print(long_words("I like to drink coffee in the morning.")) # ['morning']
print(long_words("I also have a taste for hot chocolate in the afternoon.")) # ['chocolate', 'afternoon']
print(long_words("I never drink tea late at night.")) # []


```

```python
# Splitting using Regex
import re
# It will match the literal characters and split the sentence just before that character that matches
print(re.split(r"[.?!]", "One sentence. Another one? And the last one!"))
print(re.split(r"[.?!]", "One sentence, Another one? And the last one"))

# If we include the capturing group enclosing [] then it will include the literal characters as well
print(re.split(r"([.?!])", "One sentence. Another one? And the last one!"))

# Sub method in RE module is used to substitue the text
    # the first [] will cover \w = all alphanumeric and _ the others are literal characters the second [] will cover \w = all alphanumeric and . the others are literal characters
print(re.sub(r"[\w.%+-]+@[\w.-]+", "[REDACTED]", "Received an email for go_nuts95@my.example.com"))

```

```python
import re
def transform_record(record):
  new_record = re.sub((r"(\d{3}-\d{3}-\d{4}|\d{3}-\d)"), r"+1-\1", record)
  return new_record

print(transform_record("Sabrina Green,802-867-5309,System Administrator")) 
# Sabrina Green,+1-802-867-5309,System Administrator

print(transform_record("Eli Jones,684-3481127,IT specialist")) 
# Eli Jones,+1-684-3481127,IT specialist

print(transform_record("Melody Daniels,846-687-7436,Programmer")) 
# Melody Daniels,+1-846-687-7436,Programmer

print(transform_record("Charlie Rivera,698-746-3357,Web Developer")) 
# Charlie Rivera,+1-698-746-3357,Web Developer
```

```python
# Find the regular expression for consecutive vowel
import re
def multi_vowel_words(text):
  # Initial \w* = any word character previous to the matched pattern inside [] bracket (That will include all the letters in that word before the match)
  # Second \w* = Any word character after the patterns are matched, then it will ensure to complete the word
  # * will make sure the entire word is printed
  pattern = r"\w*[aeiouAEIOU]{3,}\w*"
  result = re.findall(pattern, text)
  return result

print(multi_vowel_words("Life is beautiful")) 
# ['beautiful']

print(multi_vowel_words("Obviously, the queen is courageous and gracious.")) 
# ['Obviously', 'queen', 'courageous', 'gracious']

print(multi_vowel_words("The rambunctious children had to sit quietly and await their delicious dinner.")) 
# ['rambunctious', 'quietly', 'delicious']

print(multi_vowel_words("The order of a data queue is First In First Out (FIFO)")) 
# ['queue']

print(multi_vowel_words("Hello world!")) 
# []
```

```python
import re
def transform_comments(line_of_code):
  # here first capture group takes # out, and then second capture group takes everything apart new line character
  result = re.sub(r"(#+\s*)(.+)", r"// \2", line_of_code)
  return result


print(transform_comments("### Start of program")) 
# Should be "// Start of program"
print(transform_comments("  number = 0   ## Initialize the variable")) 
# Should be "  number = 0   // Initialize the variable"
print(transform_comments("  number += 1   # Increment the variable")) 
# Should be "  number += 1   // Increment the variable"
print(transform_comments("  return(number)")) 
# Should be "  return(number)"

print(re.search(r"#+\s*", "### Start of program"))
```

```python
import re
import csv
def contains_domain(address, domain):
  """Returns True if the email address contains the given,domain,in the domain position, false if not."""
  domain = r'[\w\.-]+@'+domain+'$'
  if re.match(domain,address):
    return True
  return False
def replace_domain(address, old_domain, new_domain):
  """Replaces the old domain with the new domain in the received address."""
  old_domain_pattern = r'' + old_domain + '$'
  address = re.sub(old_domain_pattern, new_domain, address)
  return address
def main():
  """Processes the list of emails, replacing any instances of the old domain with the new domain."""
  old_domain, new_domain = 'abc.edu', 'xyz.edu'
  csv_file_location = '/Users/vishv/Coursera_IT_Automation/2_Python_to_interact_with_OS/3_regular_expressions/user_emails.csv'
  report_file = '/Users/vishv/Coursera_IT_Automation/2_Python_to_interact_with_OS/3_regular_expressions' + '/updated_user_emails.csv'
  user_email_list = []
  old_domain_email_list = []
  new_domain_email_list = []
  with open(csv_file_location, 'r') as f:
    user_data_list = list(csv.reader(f))
    user_email_list = [data[1].strip() for data in user_data_list[1:]]
    for email_address in user_email_list:
     if contains_domain(email_address, old_domain):
        old_domain_email_list.append(email_address)
        replaced_email = replace_domain(email_address,old_domain,new_domain)
        new_domain_email_list.append(replaced_email)
    email_key = ' ' + 'Email Address'
    email_index = user_data_list[0].index(email_key)
    for user in user_data_list[1:]:
      for old_domain, new_domain in zip(old_domain_email_list, new_domain_email_list):
        if user[email_index] == ' ' + old_domain:
          user[email_index] = ' ' + new_domain
  f.close()
  with open(report_file, 'w+') as output_file:
    writer = csv.writer(output_file)
    writer.writerows(user_data_list)
    output_file.close()


main()






```

```python
def find_productID(report):
  pattern = r"\b([1]{1}\d{3}-[A-Z]{2}-\d{2})\b"  #enter the regex pattern here
  result = re.findall(pattern, report) #enter the re method  here
  return result
  
print(find_productID("Products 1234-AB-30 and 2234-AB-30, not items 12-AB-30 or 12345-AB-30")) # Should return ['1234-AB-30']
print(find_productID("Products of interest are 1234-AB-30, 1678-XZ-11, and 1561-CD-57. We're not interested in other products like 2345-AB-29.")) # Should return ['1234-AB-30', '1678-XZ-11', '1561-CD-57']
```
