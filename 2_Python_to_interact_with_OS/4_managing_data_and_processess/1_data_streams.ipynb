$PATH = It is an env. variable that tells the shell where to look for
        the executable files when we type any command in CLI

    shell ----> $PATH --------> hey executables are in this dir
    ---> shell executes them and run the programs.

If command is not found ----> export the executable file path into
$PATH env. Variable -----> restart terminal -----> now env variable 
will tell CLI where to look for that command/executable file.
