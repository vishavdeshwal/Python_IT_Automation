```python
import subprocess
subprocess.run(["sleep", "2"])
subprocess.run(["date"])
subprocess.run(["ls"])
```

```python
import subprocess

# I'm not capturing the output of the subprocess, it will directly printed in terminal
result = subprocess.run(["ls", "this_dir"])
print(result)
print(result.returncode)
```

Obtaining the output of a system command

```python
import subprocess

# I'm capturing the output of the subprocess, it means it will not print in the terminal
# I can use the stdout attribute and manipulate the program accordingly, since I have the output for the command is captured.
result = subprocess.run(["host", "8.8.8.8"], capture_output=True)
print(result.stdout) # ----> initial b at the fron tells that it is an array of bytes and not string
print(result.returncode)

# So to change the output to string we will use decode() function
# Unicode standard  UTF-8 is used to convert an array of bytes into strings
print(result.stdout.decode().split())
```

```python
import subprocess

result = subprocess.run(["rm", "this_dir"], capture_output=True)
print(result.returncode)
print(result.stdout) # it will be empty because rm is a command could not be executed properly
print(result.stderr) # Now this will give reson for why rm command was not successfully executed
```

```python
# More advanced subject

import os
import subprocess

# This will copy all the environment variable into this dictionary variable my_env
my_env = os.environ.copy()
# Now this will add /opt/myapp/ to the PATH variable(key) inside the list of dict my_env and append
my_env["PATH"] = os.pathsep.join(["/opt/myapp/", my_env["PATH"]])

result = subprocess.run(["myapp"], env=my_env)
```
